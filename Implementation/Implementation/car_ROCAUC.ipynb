```python
import pandas as pd 
from sklearn.model_selection import train_test_split, cross_val_predict
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder
from sklearn.metrics import roc_curve, auc
from sklearn.metrics import roc_auc_score
from sklearn.linear_model import RidgeClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics
from sklearn import svm
from sklearn import tree
from yellowbrick.classifier import ROCAUC
```

```python
data=pd.read_csv("balanced_car_rand.csv")
```

```python
data.columns=["id","buying","Oxin","Onine","Hormone","Tsh","Resin","Class"]
```

```python
data.set_index("id", inplace=True)
```

```python
df1 = data[['Class']]
```

```python
df2 = data[["buying","Oxin","Onine","Hormone","Tsh","Resin","Class"]]
```

```python
data_class=data.Class
```

```python
X = OrdinalEncoder().fit_transform(df2)
y = LabelEncoder().fit_transform(data_class)
```

```python
#X
```

```python
#y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
model=svm.SVC(C = 1, probability=True)
clf = DecisionTreeClassifier(random_state=0, max_depth=3)
knn = KNeighborsClassifier(n_neighbors=5)
```

```python
my_title = "ROCAUC Curves for DecisionTreeClassifier on car"
#title=my_title
```

```python
visualizer = ROCAUC(model, classes=["Unacc", "Acc", "Good", "Very Good"])
visualizer_1 = ROCAUC(clf, classes=["Unacc", "Acc", "Good", "Very Good"], title=my_title)
visualizer_2 = ROCAUC(knn, classes=["Unacc", "Acc", "Good", "Very Good"])
```

```python
visualizer.fit(X_train, y_train)        # Fit the training data to the visualizer
visualizer.score(X_test, y_test)        # Evaluate the model on the test data
visualizer.show()        
```

```python
visualizer_1.fit(X_train, y_train)        # Fit the training data to the visualizer
visualizer_1.score(X_test, y_test)        # Evaluate the model on the test data
visualizer_1.show()        
```

```python
visualizer_2.fit(X_train, y_train)        # Fit the training data to the visualizer
visualizer_2.score(X_test, y_test)        # Evaluate the model on the test data
visualizer_2.show()    
```
